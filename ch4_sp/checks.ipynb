## (a) LP with fixed $Q$ equals the average of `profit` on the same scenarios

```python
_, R, D, SP = generate(200, 1, inst)
Q = np.full(10, 40.0)
v_lp = solve_saa(R, D, SP, inst, Q_fixed=Q)[1]
v_pf = profit(Q, R, D, SP, inst).mean()
print(f"LP = {v_lp:.6f}, profit = {v_pf:.6f}, diff = {abs(v_lp - v_pf):.1e}")
```
